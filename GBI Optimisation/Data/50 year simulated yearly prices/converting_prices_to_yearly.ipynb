```python
from codelib.sampling.data_sampling import yearly_price
import numpy as np
import pandas as pd
from codelib.file_management.dynamic_file_pathing import get_root
import os
```

```python
root = get_root()
output_folder = os.path.join(root, "GBI Optimisation", "Data", "50 year simulated yearly prices")
data_folder = r"/Users/osito/Library/CloudStorage/OneDrive-SharedLibraries-CBS-CopenhagenBusinessSchool/Thesis - General/Simulation/Asset Prices/50y"

data_dictionary = {
    "Ejendomme": "Ejendomme_50y.xlsx",
    "Emerging markets aktier": "Emerging markets aktier_50y.xlsx",
    "Emerging markets statsobligationer": "Emerging markets statsobligationer_50y.xlsx",
    "Globale aktier (developed markets)": "Globale aktier (developed markets)_50y.xlsx",
    "Hedgefonde": "Hedgefonde_50y.xlsx",
    "High-yield obligationer": "High-yield obligationer_50y.xlsx",
    "Infrastruktur": "Infrastruktur_50y.xlsx",
    "Investment-grade obligationer": "Investment-grade obligationer_50y.xlsx",
    "Private equity": "Private equity_50y.xlsx",
    "Stats- og realkreditobligationer": "Stats- og realkreditobligationer_50y.xlsx"
}

for name, file in data_dictionary.items():
    print(f"Initiating {name}")
    path = os.path.join(data_folder, file)
    data = pd.read_excel(path)
    yearly_data = yearly_price(data, 52,)

    output_path = os.path.join(output_folder, f"{name}_yearly.xlsx")
    yearly_data.to_excel(output_path, index=False, header=True)

    print(f"{name} complete\n")

print("All conversions complete")
```
