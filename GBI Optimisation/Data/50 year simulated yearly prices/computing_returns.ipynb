```python
import os
import subprocess
import pandas as pd
from codelib.file_management.dynamic_file_pathing import get_root
```

```python
root = get_root()
data_folder=os.path.join(root, "GBI Optimisation", "Data", "50 year simulated yearly prices")
output_folder=os.path.join(root, "GBI Optimisation", "Data", "50 year simulated yearly prices")
```

```python
file_directory = {
    "EM Aktier": "Emerging markets aktier_yearly.xlsx",
    "EM Statsobl.": "Emerging markets statsobligationer_yearly.xlsx",
    "Globale Aktier": "Globale aktier (developed markets)_yearly.xlsx",
    "HY Obl.": "High-yield obligationer_yearly.xlsx",
    "IG Obl.": "Investment-grade obligationer_yearly.xlsx",
    "Stats & RK Obl.": "Stats- og realkreditobligationer_yearly.xlsx"
}

total_files=len(file_directory)
for i, (name, file_name) in enumerate(file_directory.items(), start=1):
    file_path = os.path.join(data_folder, file_name)

    print(f"Computing yearly returns for [{name}] [{i}/{total_files}]")
    data=pd.read_excel(file_path)
    returns=data.pct_change(axis=1).dropna(axis=1)
    output_path=os.path.join(output_folder, f"{name}_yearly_returns.xlsx")
    returns.to_excel(output_path, index=False, header=True)
    print(f"Conversion complete\n"
          f"Exported to {output_path}\n\n")

print("ALL CONVERSIONS COMPLETE")
```
