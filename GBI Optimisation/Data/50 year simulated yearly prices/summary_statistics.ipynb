```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from codelib.file_management.dynamic_file_pathing import get_root
from IPython.display import display, HTML
```

```python
root = get_root()
data_folder=os.path.join(root, "GBI Optimisation", "Data", "50 year simulated yearly prices")
output_folder=os.path.join(root, "GBI Optimisation", "Data", "50 year simulated yearly prices", "Graphs")
```

```python
file_directory = {
    "EM Aktier": "EM Aktier_yearly_returns.xlsx",
    "EM Statsobl.": "EM Statsobl._yearly_returns.xlsx",
    "Globale Aktier": "Globale Aktier_yearly_returns.xlsx",
    "HY Obl.": "HY Obl._yearly_returns.xlsx",
    "IG Obl.": "IG Obl._yearly_returns.xlsx",
    "Stats & RK Obl.": "Stats & RK Obl._yearly_returns.xlsx"
}

stats_dict = {}
```

```python
for asset_name, file_name in file_directory.items():
    file_path = os.path.join(data_folder, file_name)

    # Load returns
    df = pd.read_excel(file_path, header=0)

    # Compute summary stats
    stats = pd.DataFrame({
        'mean':     df.mean(),
        'std':      df.std(),
        'skew':     df.skew(),
        'kurtosis': df.kurtosis(),
        'median':   df.median(),
        '5%':       df.quantile(0.05),
        '95%':      df.quantile(0.95)
    })

    # Transpose & round
    stats_t = stats.T.round(4)

    # Store for later
    stats_dict[asset_name] = stats_t

    # Display
    display(HTML(f"<h3>{asset_name}</h3>"))
    display(
        stats_t
        .style
        .format("{:.4f}")
        .background_gradient(axis=1)
    )
```

```python
mean_returns = stats_t.loc['mean']

plt.figure(figsize=(8, 4))
plt.plot(mean_returns.index, mean_returns.values, linestyle='-')
plt.xlabel('Year')
plt.ylabel('Mean Return')
plt.title('Mean Return over Years')
plt.grid(True)
plt.xticks(rotation=45)      # rotate labels if they’re long (e.g. 'Year 1', 'Year 2', …)
plt.tight_layout()
plt.show()
```

```python
for asset_name, stats_t in stats_dict.items():
    # 1. Extract the mean-return series
    mean_returns = stats_t.loc['mean']

    # 2. Plot
    plt.figure(figsize=(8, 4))
    plt.plot(
        mean_returns.index,
        mean_returns.values,
        marker='o',
        linestyle='-'
    )
    plt.title(f'Mean Return over Years: {asset_name}')
    plt.xlabel('Year')
    plt.ylabel('Mean Return')
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()

    # 3. Save to file (optional)
    fname = f"{asset_name.replace(' ', '_')}_mean_return.png"
    plt.savefig(os.path.join(output_folder, fname), dpi=300)

    # 4. Show
    plt.show()
```

```python
for asset_name, stats_t in stats_dict.items():
    # Extract series
    years = stats_t.columns
    mean_ret = stats_t.loc['mean']
    lower_ci = stats_t.loc['5%']
    upper_ci = stats_t.loc['95%']

    # Plot
    plt.figure(figsize=(8, 4))
    plt.plot(years, mean_ret.values, marker='o', linestyle='-', label='Mean')
    plt.fill_between(years, lower_ci.values, upper_ci.values, alpha=0.2, label='5th–95th percentile')

    plt.title(f'{asset_name}: Mean Return with 90% CI')
    plt.xlabel('Year')
    plt.ylabel('Return')
    plt.xticks(rotation=45)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    # Save and show
    fname = f"{asset_name.replace(' ', '_')}_mean_CI.png"
    plt.savefig(os.path.join(output_folder, fname), dpi=300)
    plt.show()
```

```python
plt.figure(figsize=(10, 6))

for asset_name, stats_t in stats_dict.items():
    # Extract the mean-return series
    mean_returns = stats_t.loc['mean']

    # Plot each asset’s curve
    plt.plot(
        mean_returns.index,
        mean_returns.values,
        marker='o',
        linestyle='-',
        label=asset_name
    )

# Final touches
plt.title('Expected Return Trajectories by Asset')
plt.xlabel('Year')
plt.ylabel('Mean Return')
plt.xticks(rotation=45)    # rotate labels if they’re long
plt.grid(True)
plt.legend(title='Asset')
plt.tight_layout()

# Show the combined plot
plt.show()
```
