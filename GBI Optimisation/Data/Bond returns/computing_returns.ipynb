```python
import os
import subprocess
import pandas as pd
from codelib.file_management.dynamic_file_pathing import get_root
```

```python
root = get_root()
data_folder=os.path.join(root, "GBI Optimisation", "Data", "Bond returns")
output_folder=os.path.join(root, "GBI Optimisation", "Data", "Bond returns")
data_path=os.path.join(data_folder, "bond_prices_yearly.xlsx")
data= pd.read_excel(data_path)
```

```python
returns = data.pct_change(axis=1).dropna(axis=1)
```

```python
output_path=os.path.join(output_folder, "bond_prices_yearly_returns.xlsx")
returns.to_excel(output_path, index=False, header=True)
```
