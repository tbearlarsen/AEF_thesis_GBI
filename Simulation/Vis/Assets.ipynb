# Visualising Simulated Data

```python
import pandas as pd
import numpy as np
from codelib.file_management.dynamic_file_pathing import get_root
import os
import matplotlib.pyplot as plt
from codelib.Johan.visualization.base import fan_chart
import seaborn as sns
```

## Load data

```python
root = get_root()
data_path = os.path.join(root, "GBI Optimisation", "Data", "50 year simulated yearly prices", "EM Aktier_yearly_returns.xlsx")
data = pd.read_excel(data_path, header = 0)
```

```python
# Extract the last column
last_column = data.iloc[:, -1]

# Plot the distribution
fig, ax = plt.subplots(figsize=(10, 6))
sns.kdeplot(last_column, fill=True, color='navy', ax=ax)

ax.set_title('Distribution of returns t=50')
ax.set_xlabel('Returns')
ax.set_ylabel('Density')

plt.show()
```
